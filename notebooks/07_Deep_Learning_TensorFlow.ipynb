# Deep Learning with TensorFlow

This notebook implements the TensorFlow portion of the handbook with tensors, a neural network, compilation, training, evaluation metrics and prediction.

## 1. Import TensorFlow

```python
import tensorflow as tf

print("TensorFlow version:", tf.__version__)
```

## 2. Tensors

The handbook describes tensors as the fundamental units of data in TensorFlow.

```python
x = tf.constant([[1.0, 2.0], [3.0, 4.0]])
print(x)
print("Shape:", x.shape)
```

## 3. Create Original Training Data

```python
import numpy as np

rng = np.random.default_rng(42)

X = rng.normal(size=(200, 2)).astype("float32")
y = ((X[:, 0] + X[:, 1]) > 0).astype("float32")

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## 4. Build a Neural Network

TensorFlow/Keras provides a high-level way to define neural networks.

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(2,)),
    tf.keras.layers.Dense(16, activation="relu"),
    tf.keras.layers.Dense(8, activation="relu"),
    tf.keras.layers.Dense(1, activation="sigmoid")
])

model.summary()
```

## 5. Compile the Model

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

## 6. Train the Model

```python
history = model.fit(
    X,
    y,
    epochs=20,
    batch_size=16,
    validation_split=0.2,
    verbose=0
)

print("Final training accuracy:", history.history["accuracy"][-1])
print("Final validation accuracy:", history.history["val_accuracy"][-1])
```

## 7. Prediction

```python
sample = np.array([[1.0, 1.0], [-1.0, -1.0]], dtype="float32")
probabilities = model.predict(sample, verbose=0)

print(probabilities)
```
