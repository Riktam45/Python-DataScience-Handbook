# Data Analysis and Preprocessing

This notebook follows the handbook's data-analysis workflow and includes inspection, cleaning, EDA, visualization, missing-value handling, and outlier detection.

## 1. Define the Objective

A data-analysis workflow starts by defining the question, collecting data, cleaning it, exploring it, analyzing it, and communicating the findings.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Create a Small Original Dataset

This notebook uses a small dataset created specifically for the handbook. No external dataset platform is required.

```python
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Age": [24, 31, 29, 42, 35, 27],
    "Salary": [45000, 62000, 58000, 90000, 72000, 51000],
    "Department": ["Data", "IT", "Data", "Management", "IT", "Data"]
})
df
```

## 3. Inspect the Data

```python
print(df.head())
print("\nShape:", df.shape)
print("\nInfo:")
df.info()

print("\nDescriptive statistics:")
print(df.describe(numeric_only=True))
```

## 4. Cleaning and Missing Values

```python
df_clean = df.copy()
df_clean.loc[2, "Salary"] = np.nan

print("Missing values:")
print(df_clean.isna().sum())

print("\nRows after dropping missing values:")
print(df_clean.dropna())

print("\nFilling missing salary with the mean:")
df_filled = df_clean.copy()
df_filled["Salary"] = df_filled["Salary"].fillna(df_filled["Salary"].mean())
print(df_filled)
```

## 5. Exploratory Data Analysis

```python
print("Average salary:", df_filled["Salary"].mean())
print("Median salary:", df_filled["Salary"].median())
print("Minimum salary:", df_filled["Salary"].min())
print("Maximum salary:", df_filled["Salary"].max())

print("\nAverage salary by department:")
print(df_filled.groupby("Department")["Salary"].mean())
```

## 6. Visualization During EDA

```python
sns.boxplot(data=df_filled, x="Department", y="Salary")
plt.title("Salary by Department")
plt.show()

sns.scatterplot(data=df_filled, x="Age", y="Salary", hue="Department")
plt.title("Age vs Salary")
plt.show()
```

## 7. Outlier Detection with Z-Score

The handbook discusses Z-score-based outlier detection. A commonly used threshold is an absolute Z-score greater than 3.

```python
values = df_filled["Salary"].to_numpy()
mean = values.mean()
std = values.std()

z_scores = (values - mean) / std
outliers = df_filled[np.abs(z_scores) > 3]

print("Z-scores:", z_scores)
print("\nPotential outliers:")
print(outliers)
```

## 8. Outlier Detection with IQR

The handbook also uses the Interquartile Range method:
`IQR = Q3 - Q1`.

Values below `Q1 - 1.5 × IQR` or above `Q3 + 1.5 × IQR` are treated as outliers.

```python
data = np.array([6, 2, 3, 4, 5, 1, 50])

Q1 = np.percentile(data, 25)
Q2 = np.percentile(data, 50)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

low_lim = Q1 - 1.5 * IQR
up_lim = Q3 + 1.5 * IQR

outlier = [x for x in data if x < low_lim or x > up_lim]

print("Q1:", Q1)
print("Median:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", low_lim)
print("Upper limit:", up_lim)
print("Outliers:", outlier)

sns.boxplot(x=data)
plt.show()
```
