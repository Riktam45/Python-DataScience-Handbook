# Data Visualization with Matplotlib and Seaborn

This notebook converts the handbook's visualization concepts into runnable examples using Matplotlib and Seaborn.

## 1. Imports and Example Data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 135, 180, 210, 195],
    "Profit": [20, 32, 28, 40, 55, 48],
    "Category": ["A", "B", "A", "B", "A", "B"]
})

df
```

## 2. Line Plot

```python
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

## 3. Bar Chart

```python
plt.bar(df["Month"], df["Sales"])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

## 4. Scatter Plot

```python
plt.scatter(df["Sales"], df["Profit"])
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

## 5. Histogram

```python
plt.hist(df["Sales"], bins=5)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

## 6. Seaborn Scatter Plot

```python
sns.scatterplot(data=df, x="Sales", y="Profit", hue="Category")
plt.title("Sales vs Profit by Category")
plt.show()
```

## 7. Seaborn Box Plot

```python
sns.boxplot(data=df, y="Sales")
plt.title("Sales Distribution")
plt.show()
```

## 8. Heatmap

```python
numeric = df[["Sales", "Profit"]]
sns.heatmap(numeric.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
