# Deep Learning with PyTorch

This notebook turns the handbook's PyTorch material into a runnable neural-network example using tensors, a model, loss function, optimizer, forward pass, backpropagation and a training loop.

## 1. Import PyTorch

```python
import torch
import torch.nn as nn
import torch.optim as optim

print("PyTorch version:", torch.__version__)
```

## 2. Tensors

```python
x = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
print(x)
print("Shape:", x.shape)
```

## 3. Create Dummy Training Data

The handbook demonstrates a training loop using 100 samples and 10 features.

```python
torch.manual_seed(42)

inputs = torch.randn((100, 10))
targets = torch.randint(0, 2, (100, 1)).float()

print("Inputs shape:", inputs.shape)
print("Targets shape:", targets.shape)
```

## 4. Define a Neural Network

PyTorch neural networks are built with `torch.nn.Module`.

```python
class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(10, 16)
        self.activation = nn.ReLU()
        self.layer2 = nn.Linear(16, 1)
        self.output = nn.Sigmoid()

    def forward(self, x):
        x = self.layer1(x)
        x = self.activation(x)
        x = self.layer2(x)
        return self.output(x)

model = NeuralNetwork()
print(model)
```

## 5. Loss Function and Optimizer

```python
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

## 6. Training Loop

The handbook describes the sequence as:
1. clear accumulated gradients,
2. forward pass,
3. compute loss,
4. backpropagation,
5. update weights.

```python
epochs = 20

for epoch in range(epochs):
    optimizer.zero_grad()
    outputs = model(inputs)
    loss = criterion(outputs, targets)
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 5 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")
```

## 7. Prediction

```python
with torch.no_grad():
    probabilities = model(inputs[:5])
    predictions = (probabilities >= 0.5).float()

print("Probabilities:")
print(probabilities)
print("\nPredictions:")
print(predictions)
```
