# Pandas — Data Manipulation

This notebook covers the Pandas concepts presented in the handbook: Series, DataFrames, indexing, filtering, missing values, date/time filtering, file I/O, GroupBy, and aggregation.

## 1. Importing Pandas

```python
import pandas as pd
import numpy as np
```

## 2. Series

```python
sales = [1500, 2200, 1850, 2900, 3100]
sales_series = pd.Series(sales)
print(sales_series)

temps = [25.5, 27.1, 23.8, 28.0]
days = ["Mon", "Tue", "Wed", "Thu"]
temp_series = pd.Series(temps, index=days)
print("\nTemperature Series:")
print(temp_series)

data_dict = {"p": 0, "y": 1, "t": 2, "h": 3, "o": 4, "n": 5}
print("\nSeries from dictionary:")
print(pd.Series(data_dict))
```

## 3. DataFrames

```python
employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve"]
}
df = pd.DataFrame(employee_data)
print(df)

data = [
    ["New York", 12.0, 8.4],
    ["London", 11.5, 8.9],
]
df_list = pd.DataFrame(data, columns=["City", "Avg_Temp_C", "Population_M"])
print("\nDataFrame from list:")
print(df_list)

data_array = np.array([
    [180, 75, 25],
    [165, 60, 30],
    [175, 70, 45],
    [188, 85, 22]
])
df_numpy = pd.DataFrame(
    data_array,
    columns=["Height_cm", "Weight_kg", "Age"],
    index=["A", "B", "C", "D"]
)
print("\nDataFrame from NumPy array:")
print(df_numpy)
```

## 4. Indexing and Accessing Data

```python
data = {
    "Name": ["John", "Alice", "Bob", "Eve", "Charlie"],
    "Age": [25, 30, 22, 35, 28],
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "Salary": [50000, 55000, 40000, 70000, 48000]
}
df = pd.DataFrame(data)

print("Index:", df.index)
print("\nAge column:")
print(df["Age"])

print("\nSecond row with iloc:")
print(df.iloc[1])
```

### `loc` and `iloc`

```python
df_named = df.set_index("Name")

print("Row by label:")
print(df_named.loc["Alice"])

print("\nRows by position:")
print(df.iloc[[0, 2]])

print("\nRows and columns by position:")
print(df.iloc[[0, 2], [0, 2]])
```

## 5. Filtering

```python
sample = pd.DataFrame({
    "Name": ["RACHEL", "MONICA", "PHOEBE", "ROSS", "CHANDLER", "JOEY"],
    "Age": [30, 35, 37, 33, 34, 30],
    "Salary": [100000, 93000, 88000, 120000, 94000, 95000],
    "JOB": ["DESIGNER", "CHEF", "MASUS", "PALEONTOLOGY", "IT", "ARTIST"]
})

print("Age > 30:")
print(sample[sample["Age"] > 30])

print("\nAge > 30 and Salary > 100000:")
print(sample[(sample["Age"] > 30) & (sample["Salary"] > 100000)])

print("\nJOB is IT or CHEF:")
print(sample[sample["JOB"].isin(["IT", "CHEF"])])

print("\nJOB starts with A:")
print(sample[sample["JOB"].str.startswith("A")])

print("\nAge between 30 and 35:")
print(sample[sample["Age"].between(30, 35)])

print("\nSQL-like query:")
print(sample.query("Age > 30 and Salary < 100000"))
```

## 6. Missing Values and Dates

```python
missing_df = sample.copy()
missing_df.loc[2, "Salary"] = np.nan

print("Missing salary:")
print(missing_df[missing_df["Salary"].isna()])

print("\nAfter dropna:")
print(missing_df.dropna())

date_df = pd.DataFrame({
    "date": pd.to_datetime(["2024-05-10", "2024-05-20", "2024-06-15", "2023-05-12"]),
    "value": [10, 20, 30, 40]
})

print("\nMay 2024:")
print(date_df[(date_df["date"].dt.year == 2024) & (date_df["date"].dt.month == 5)])
```

## 8. GroupBy and Aggregation

```python
scores = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Subject": ["Python", "Python", "Python", "Math", "Math", "Math"],
    "Score": [85, 92, 78, 88, 95, 81]
})

print("Mean score per subject:")
print(scores.groupby("Subject")["Score"].mean())

print("\nMultiple aggregations:")
print(scores.groupby("Subject")["Score"].agg(["mean", "max", "min", "count"]))

print("\nDescribe per subject:")
print(scores.groupby("Subject")["Score"].describe())
```
